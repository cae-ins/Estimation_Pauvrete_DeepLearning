```python
import pandas as pd
import os

# Chemins
csv_path = r"D:\wealth_predict_sentinel\Data\processed_csv\df_final_afrique_pharmacies_schools_combined_sampled_class_optimal_classified_8k_obs.csv"
images_path =r"D:\wealth_predict_sentinel\Data\downloaded\Image_satellite_Divers_Pays_Afrique_Zoom_14_Sentinel_2_an_2023"

# Charger le DataFrame
data = pd.read_csv(csv_path)

# Fonction pour générer le nom de l'image
def get_image_name(row):
    latitude = row['latitude']
    longitude = row['longitude']
    # Format du nom de l'image
    image_name = f"lat_{latitude}_lon_{longitude}_Zoom_14.jpeg"
    # Vérifier si l'image existe
    image_path = os.path.join(images_path, image_name)
    return image_name if os.path.exists(image_path) else None

# Ajouter la colonne 'Image_Name'
data['Image_Name'] = data.apply(get_image_name, axis=1)

# Réorganiser les colonnes
data = data[['Image_Name', 'latitude', 'longitude', 'radiance_16_km^2', 'class']]

# Sauvegarder le nouveau DataFrame
output_path = r"D:\wealth_predict_sentinel\Data\processed_csv\Data_Africa_sampled_with_images.csv"
data.to_csv(output_path, index=False)

print(f"Nouveau fichier sauvegardé à : {output_path}")
```
